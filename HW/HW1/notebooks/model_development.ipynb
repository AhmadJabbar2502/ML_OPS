```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
df = pd.read_csv("../data/sampregdata.csv")

df.head()
```

```python
X_columns = ["x1", "x2", "x3", "x4"]
y = df["y"]

results = {}

for column in X_columns:
    X = df[[column]]

    model = LinearRegression()
    model.fit(X, y)

    predictions = model.predict(X)
    r2 = r2_score(y, predictions)

    results[column] = r2

results
```

```python
results_df = pd.DataFrame(
    list(results.items()),
    columns=["X Variable", "R²"]
).sort_values("R²", ascending=False)

results_df
```

```python
X = df[["x4"]]
y = df["y"]

model_v1 = LinearRegression()
model_v1.fit(X, y)

print("Intercept:", model_v1.intercept_)
print("Coefficient:", model_v1.coef_[0])
print("R²:", model_v1.score(X, y))
```

```python
import joblib

joblib.dump(model_v1, "../models/model_v1.pkl")
```
